**Мова програмування та середовище:** Python 3, Google Colab


```python
import matplotlib.pyplot as plt
from google.colab import files
```

```python
file_path = 'DS0.txt'
```

```python
x_coords = []
y_coords = []
```

```python
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) >= 2:
            x, y = int(parts[0]), int(parts[1])
            x_coords.append(x)
            y_coords.append(y)

print(f"Зчитано точок: {len(x_coords)}")
print(f"X: від {min(x_coords)} до {max(x_coords)}")
print(f"Y: від {min(y_coords)} до {max(y_coords)}")
```

```python
dpi = 100
width_px, height_px = 960, 540

fig = plt.figure(figsize=(width_px / dpi, height_px / dpi), dpi=dpi)
ax = fig.add_axes([0, 0, 1, 1])
ax.set_xlim(0, width_px)
ax.set_ylim(0, height_px)

ax.axis('off')
fig.patch.set_facecolor('white')
ax.scatter(x_coords, y_coords, s=2, color='black')
output_filename = 'result_DS0.png'
plt.savefig(output_filename, dpi=dpi)
plt.show()
```
